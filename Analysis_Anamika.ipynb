### **Identifying Salary Outliers**

The boxplot suggests that there are several unusually high salary values. I used the Interquartile Range (IQR) method to calculate the lower and upper bounds for potential outliers.


$$IQR = Q_3 - Q_1$$


$$\text{Lower Bound} = Q_1 - 1.5(IQR)$$


$$\text{Upper Bound} = Q_3 + 1.5(IQR)$$


```python
# 13. Calculating IQR and outlier bounds for median annual salary

Q1 = df["median_annual_salary"].quantile(0.25)
Q3 = df["median_annual_salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_summary = pd.DataFrame({
    "Statistic": ["Q1", "Q3", "IQR", "Lower Bound", "Upper Bound"],
    "Value": [Q1, Q3, IQR, lower_bound, upper_bound]
})

outlier_summary
```

```python
# 14. Counting potential salary outliers

salary_outliers = df[
    (df["median_annual_salary"] < lower_bound) |
    (df["median_annual_salary"] > upper_bound)
]

pd.DataFrame({
    "Number of Potential Outliers": [len(salary_outliers)]
})
```

```python
# 15. Viewing the highest salary outliers

salary_outliers[
    ["occupational_title", "median_annual_salary"]
].sort_values(
    "median_annual_salary",
    ascending=False
).head(10)
```

#### **Finding:**  

*The IQR method identified 2,263 potential salary outliers. The upper outlier threshold is approximately **`$97,793`**, and the highest salaries in the dataset exceed **`$220,000`**.*

*The highest values are associated with occupations such as Nurse Anesthetists and Musicians & Singers. Because these values correspond to real occupations and may reflect differences across geographic areas, I will not automatically remove them. Instead, I will keep them unless later analysis suggests that they are data errors or inappropriate for the comparison.*

## 3. Education Level vs. Salary

To explore my hypothesis, I compared median annual salaries across the different typical education requirement categories.

```python
# 16. Median annual salary by education requirement

salary_by_education = (
    df.groupby("typical_education_requirement")["median_annual_salary"]
    .median()
    .sort_values()
    .reset_index()
)

salary_by_education.columns = [
    "Education Requirement",
    "Median Annual Salary"
]

salary_by_education
```

#### **Finding:**  

*The difference between the highest and lowest group medians is:*

$$ 79{,}572 - 33{,}544 = 46{,}028 $$

*So the median salary for occupations requiring a doctoral or professional degree is about **`$46,028`** higher than for occupations requiring no formal educational credential.*

### Median Salary by Education Requirement

The median salary for each education category provides a simple way to compare typical earnings across different education levels.

```python
# 17. Bar chart of median annual salary by education requirement

plt.figure(figsize=(12, 6))

sns.barplot(
    data=salary_by_education,
    y="Education Requirement",
    x="Median Annual Salary",
    hue="Education Requirement",
    palette="Paired",
    legend=False,
    edgecolor="black"
)

plt.title(
    "Median Annual Salary by Education Requirement",
    fontweight="bold",
    fontsize=16
)

plt.xlabel(
    "Median Annual Salary ($)",
    color="#2114B5",
    fontweight="bold",
    fontsize=16
)

plt.ylabel(
    "Education Requirement",
    color="#2114B5",
    fontweight="bold",
    fontsize=16
)

plt.show()
```

#### **Finding:**  

*The bar chart shows a clear upward pattern in median annual salary as education requirements increase. Occupations requiring lower levels of formal education generally have lower median salaries, while occupations requiring bachelor's, master's, doctoral, or professional degrees have noticeably higher median salaries.*

*The highest median salary is for occupations requiring a Doctoral or Professional Degree at about **`$79,572`**, compared with about **`$33,544`** for occupations requiring No Formal Educational Credential. This visual pattern provides initial support for my hypothesis that higher education requirements are associated with higher occupational salaries.*

### Salary Distribution by Education Requirement

The median salary comparison shows an overall increase across education levels. I also used a boxplot to examine the variation and overlap in salaries within each education requirement category.

```python
# 18. Boxplot of median annual salary by education requirement

plt.figure(figsize=(12, 7))

sns.boxplot(
    data=df,
    y="typical_education_requirement",
    x="median_annual_salary",
    hue="typical_education_requirement",
    palette="Paired",
    legend=False,
    showfliers=False
)

plt.title(
    "Salary Distribution by Education Requirement",
    fontweight="bold",
    fontsize=16
)

plt.xlabel(
    "Median Annual Salary ($)",
    color="#2114B5",
    fontweight="bold",
    fontsize=16
)

plt.ylabel(
    "Education Requirement",
    color="#2114B5",
    fontweight="bold",
    fontsize=16
)

plt.show()
```

#### **Finding:**  

*The boxplot shows that occupations with higher education requirements generally have higher salary distributions. Doctoral or Professional Degree and Master's Degree occupations have the highest salary ranges and medians, while occupations requiring No Formal Educational Credential, Some College, or a High School Diploma tend to have lower salaries.*

*There is still noticeable overlap between education groups, especially among Associate Degree, Bachelor's Degree, and Master's Degree occupations. This means education requirement is related to salary, but it is not the only factor affecting occupational earnings.*
